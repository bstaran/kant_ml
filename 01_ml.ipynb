```python
import pandas as pd
import numpy as np
import seaborn as sns

print("pandas:", pd.__version__)
print("numpy:", np.__version__)
print("seaborn:", sns.__version__)
```

```python
df = sns.load_dataset("titanic")
df.head(5)
```

```python
# 데이터셋의 크기 조회
df.shape
```

```python
# 자료형 조회
df.info()
```

```python
df.describe()
```

```python
# 각 셀의 널 값 개수 조회
df.isnull().sum()
```

```python
# 나이가 비어있는 승객의 나이, 클래스, 성별 조회
empty_age = df["age"].isnull()
print("나이 빈칸:", empty_age.sum(), "명")
df.loc[empty_age, ["age", "class", "sex"]].head()
```

```python
# 평균값, 중앙값 산출
print("평균:", round(df["age"].mean(), 2))
print("중앙값:", round(df["age"].median(), 2))

mean_age = df["age"].mean()
# fillna() : 빈칸(NaN)만 주어진 값으로 채움
df.loc[:, "age"] = df["age"].fillna(mean_age)

# df["age"].isnull().sum()
df.loc[empty_age, ["age", "class", "sex"]].head()
```

```python
df["embark_town"].value_counts(dropna=False)
```

```python
# 최빈값 추출후 빈칸에 채우는 로직
mode_town = df["embark_town"].mode()[0]
print("최빈값:", mode_town)

df.loc[:, "embark_town"] = df["embark_town"].fillna(mode_town)
df["embark_town"].value_counts(dropna=False)
```

```python
# embarked town 값으로 설정
df.loc[:, "embarked"] = df["embark_town"].str[0]
df[["embark_town", "embarked"]].isnull().sum()
```
